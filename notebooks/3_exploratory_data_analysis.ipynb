```python
#import libraries for data visualization
import matplotlib.pyplot as plt
import seaborn as sns

#create a figure with 4 subplots
fig, axes=plt.subplots(2,2,figsize=(12,8))

#fuel distribution
sns.countplot(data=df,x='fuel',ax=axes[0,0])
axes[0,0].set_title('Fuel Distribution')

#seller type distribution
sns.countplot(data=df,x='seller_type',ax=axes[0,1])
axes[0,1].set_title('Seller Type Distribution')

#transmission distribution
sns.countplot(data=df,x='transmission',ax=axes[1,0])
axes[1,0].set_title('Transmission Distribution')

#owner distribution
sns.countplot(data=df,x='owner',ax=axes[1,1])
axes[1,1].set_title('Owner Distribution')

#rotate owner labels to prevent overlapping
axes[1,1].tick_params(axis='x',rotation=20)
#adjusting spaces between plots
plt.tight_layout()
#display the figure
plt.show()
```

## 3. Numerical Feature Analysis

```python
#plotting the distribution of selling prices using histogram
sns.histplot(data=df,x='selling_price')
#add title and labels
plt.title('Distribution of Selling Prices')
plt.xlabel('Selling Price')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

```python
#plotting kilometers driven using histogram
sns.histplot(data=df,x='km_driven')
#add title and labels
plt.title('Distribution of Kilometers Driven')
plt.xlabel('Kilometers Driven')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

```python
#plot the distribution of years using histogram
sns.histplot(data=df,x='year',binwidth=1)
#add title and labels
plt.title('Distribution of Car Manufacturing Years')
plt.xlabel('Year')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

```python
#check the no.of cars for each mileage unit
df['mileage_unit'].value_counts(dropna=False)
```

```python
# Create two subplots for different mileage units
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

#mileage distribution for kmpl
sns.histplot(data=df[df['mileage_unit'] == 'kmpl'], x='mileage_value', ax=axes[0])
axes[0].set_title('Mileage Distribution (kmpl)')
axes[0].set_xlabel('Mileage (kmpl)')
axes[0].set_ylabel('Number of Cars')

#mileage distribution for km/kg
sns.histplot(data=df[df['mileage_unit'] == 'km/kg'],x='mileage_value', ax=axes[1])
axes[1].set_title('Mileage Distribution (km/kg)')
axes[1].set_xlabel('Mileage (km/kg)')
axes[1].set_ylabel('Number of Cars')

plt.tight_layout()
plt.show()
```

```python
#plot the distribution of engine capacity
sns.histplot(data=df,x='engine_value')
#add title and labels
plt.title('Distribution of Engine Capacity')
plt.xlabel('Engine Capacity (CC)')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

```python
#plot the distribution of maximum power
sns.histplot(data=df,x='max_power_value')
#add title and labels
plt.title('Distribution of Maximum Power')
plt.xlabel('Maximum Power (bhp)')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

```python
#check the no.of cars for each torque unit
df['torque_unit'].value_counts(dropna=False)
```

```python
#create two subplots for different torque units
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

#torque distribution in Nm
sns.histplot(data=df[df['torque_unit'] == 'nm'], x='torque_value', ax=axes[0])
axes[0].set_title('Torque Distribution (Nm)')
axes[0].set_xlabel('Torque (Nm)')
axes[0].set_ylabel('Number of Cars')

#torque distribution in kgm
sns.histplot(data=df[df['torque_unit'] == 'kgm'],x='torque_value', ax=axes[1])
axes[1].set_title('Torque Distribution (kgm)')
axes[1].set_xlabel('Torque (kgm)')
axes[1].set_ylabel('Number of Cars')

plt.tight_layout()
plt.show()
```

```python
#plot the distribution of seats
sns.histplot(data=df,x='seats')
#add title and labels
plt.title('Distribution of Car Seating Capacity')
plt.xlabel('Number of Seats')
plt.ylabel('Number of Cars')
#display the plot
plt.show()
```

## 4. Bivariate Analysis

```python
#relationship between manufacturing years and selling price
sns.scatterplot(data=df,x='year',y='selling_price')
#add title and labels
plt.title('Selling Price vs Manufacturing Year')
plt.xlabel('Manufacturing Year')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between kilometers driven and selling price
sns.scatterplot(data=df,x='km_driven',y='selling_price')
#add title and labels
plt.title('Selling Price vs Kilometer Driven')
plt.xlabel('Kilometer Driven')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#create two subplots to compare mileage and selling price
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

#selling price vs mileage (kmpl)
sns.scatterplot(data=df[df['mileage_unit'] == 'kmpl'], x='mileage_value', y='selling_price', ax=axes[0])
axes[0].set_title('Selling Price vs Mileage (kmpl)')
axes[0].set_xlabel('Mileage (kmpl)')
axes[0].set_ylabel('Selling Price')

#selling price vs mileage(km/kg)
sns.scatterplot(data=df[df['mileage_unit'] == 'km/kg'],x='mileage_value', y='selling_price', ax=axes[1])
axes[1].set_title('Selling Price vs Mileage (km/kg)')
axes[1].set_xlabel('Mileage (km/kg)')
axes[1].set_ylabel('Selling Price')
plt.tight_layout()
plt.show()
```

```python
#relationship between engine capacity and selling price
sns.scatterplot(data=df,x='engine_value',y='selling_price')
#add title and labels
plt.title('Selling Price vs Engine Capacity')
plt.xlabel('Engine Capacity (CC)')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between maximum power and selling price
sns.scatterplot(data=df,x='max_power_value',y='selling_price')
#add title and labels
plt.title('Selling Price vs Maximum Power')
plt.xlabel('Maximum Power (bhp)')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between number of seats and selling price
sns.boxplot(data=df,x='seats',y='selling_price')
#add title and labels
plt.title('Selling Price vs Number of Seats')
plt.xlabel('Number of Seats')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between fuel types and selling price
sns.boxplot(data=df,x='fuel',y='selling_price')
#add title and labels
plt.title('Selling Price vs Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between transmission types and selling price
sns.boxplot(data=df,x='transmission',y='selling_price')
#add title and labels
plt.title('Selling Price vs Transmission')
plt.xlabel('Transmission')
plt.ylabel('Selling price')
#display the plot
plt.show()
```

```python
#relationship between owner type and selling price
sns.boxplot(data=df,x='owner',y='selling_price')
#add title and labels
plt.title('Selling Price vs Owner Type')
plt.xlabel('Owner Type')
plt.ylabel('Selling price')
plt.xticks(rotation=20)
#display the plot
plt.show()
```

```python
#create two subplots to compare torque and selling price
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

#selling price vs torque (nm)
sns.scatterplot(data=df[df['torque_unit'] == 'nm'], x='torque_value', y='selling_price', ax=axes[0])
axes[0].set_title('Selling Price vs Torque (Nm)')
axes[0].set_xlabel('Torque (Nm)')
axes[0].set_ylabel('Selling Price')

#selling price vs torque(kgm)
sns.scatterplot(data=df[df['torque_unit'] == 'kgm'],x='torque_value', y='selling_price', ax=axes[1])
axes[1].set_title('Selling Price vs Torque (kgm)')
axes[1].set_xlabel('Torque (kgm)')
axes[1].set_ylabel('Selling Price')
plt.tight_layout()
plt.show()
```

```python
#extract the car brand from car name
df['brand']=df['name'].str.split().str[0]
#display the most common car brands
df['brand'].value_counts().head(10)
```

```python
#comparing the top 10 car brands with selling price
top_10_brands=df['brand'].value_counts().head(10).index
#filtering the dataset to include only those brands
df_top_brands=df[df['brand'].isin(top_10_brands)]

plt.figure(figsize=(12,6))
sns.boxplot(data=df_top_brands,x='brand',y='selling_price')
plt.title('Selling Price vs Top 10 Car Brands')
plt.xlabel('Car Brand')
plt.ylabel('Selling Price')
plt.xticks(rotation=45)
plt.show()
```

## 5. Correlation Heatmap

```python
#correlation heatmap for numerical columns
#select numerical columns
numeric_df=df.select_dtypes(include='number')
#calculate correlations
corr_matrix=numeric_df.corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

## 6. Summary

- Compared selling price other features using scatterplots, boxplots, histograms and a correlation heatmap.
- Selling price is right-skewed, with most cars having lower selling prices.
- New cars have higher selling prices.
- Cars with higher kilometers driven have lower selling prices.
- Maximum power, torque (Nm), engine capacity show positive relationships with selling price.
- Automatic cars have higher selling price compared to manual cars.
- First owner cars have higher selling prices.
- Maximum power has the strongest positive correlation with selling price among the numerical features.
